# 01 — Tokenizer inefficiency analysis

Visualise where the base tokenizer is wasteful on the target codebase, and which
mined candidates recover the most tokens.

Run first:
```bash
python -m src.mine_tokens configs/experiments/vocab_qlora.yaml
python -m src.analyze_tokenizer configs/experiments/vocab_qlora.yaml
```

````python
import json, sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import pandas as pd, matplotlib.pyplot as plt

cands = json.load(open('../results/candidates.json'))['candidates']
df = pd.DataFrame(cands)
df.head(20)
````

````python
# Top-30 candidates by score (frequency x current sub-token count)
top = df.sort_values('score', ascending=False).head(30)
plt.figure(figsize=(8, 8))
plt.barh(top['token'][::-1], top['score'][::-1])
plt.xlabel('score = frequency x current_subtokens')
plt.title('Highest-value candidate tokens')
plt.tight_layout()
````

````python
# Distribution of sub-tokens-per-identifier (the inefficiency we exploit)
df['current_subtokens'].plot.hist(bins=range(1, 12))
plt.xlabel('sub-tokens per candidate'); plt.title('How badly the base tokenizer fragments code identifiers')
````
