# 03 — Results

Aggregate the per-condition JSON in `results/` into the paper tables: compression,
completion accuracy across the six conditions, latency, and semantic-ID quality.

```python
import json, glob, pathlib, pandas as pd
R = pathlib.Path('../results')
def load(pattern):
    return [json.load(open(p)) for p in glob.glob(str(R / pattern))]
```

```python
# Table 1: compression ratios
rows = []
for r in load('compression_*.json'):
    rows.append({'config': r['config_name'], **r.get('reduction_pct', {}), 'bytes/token': r.get('bytes_per_token_after')})
pd.DataFrame(rows)
```

```python
# Table 2: completion accuracy across the six conditions
rows = [{'variant': r['variant'], 'EM': r['exact_match'], 'BLEU': r['bleu'],
         'edit': r['edit_ratio'], 'syntax': r['syntax_valid'], 'codebleu': r.get('codebleu')}
        for r in load('completion_*.json')]
pd.DataFrame(rows).sort_values('EM', ascending=False)
```

```python
# Table 3: latency
rows = [{'variant': r['variant'], **r['transformers']} for r in load('latency_*.json')]
pd.DataFrame(rows)
```

```python
# Table 4: semantic-ID quality
rows = []
for r in load('semantic_ids_*.json'):
    hc = r['hierarchical_consistency']; ng = r['novel_entity_generalization']
    rows.append({'config': r['config_name'], 'L1_consistency': hc['l1_prefix_consistency'],
                 'L2_consistency': hc['l2_prefix_consistency'], 'nn_lift': ng.get('lift_over_chance'),
                 'codebook_usage': r['codebook_usage']})
pd.DataFrame(rows)
```
