# 04 — Training curves & eval results
Plot the self-play reward curve and summarise the eval JSONs in `results/`.

```python
import sys, json
sys.path.insert(0, '..')
from pathlib import Path
import matplotlib.pyplot as plt
metrics_path = Path('../results/self_play_metrics.json')
if metrics_path.exists():
    m = json.loads(metrics_path.read_text())
    it = [r['iteration'] for r in m]
    plt.plot(it, [r['mean_reward'] for r in m], marker='o', label='mean')
    plt.plot(it, [r['best_reward'] for r in m], marker='^', label='best')
    plt.xlabel('self-play iteration'); plt.ylabel('episode reward'); plt.legend(); plt.title('Self-improvement curve')
else:
    print('run self-play first')
```

```python
for name in ['eval_compression', 'eval_gameplay', 'eval_planning']:
    p = Path(f'../results/{name}.json')
    if p.exists():
        print(name, '->', json.loads(p.read_text()))
```
