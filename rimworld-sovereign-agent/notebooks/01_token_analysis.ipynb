# 01 — RimWorld tokenizer inefficiency
Visualise which game terms tokenize badly and how many tokens the extended tokenizer saves.
Run knowledge extraction + `mine_tokens` first.

```python
import sys, json
sys.path.insert(0, '..')
from pathlib import Path
cands = json.loads(Path('../results/candidates.json').read_text())
rows = cands['candidates'][:30]
for r in rows:
    print(f"{r['token'][:30]:30s} freq={r.get('frequency','?'):>6} subtok={r.get('current_subtokens','?')} score={r.get('score',0):.0f}")
```

```python
import matplotlib.pyplot as plt
labels = [r['token'][:18] for r in rows[:20]]
scores = [r.get('score', 0) for r in rows[:20]]
plt.figure(figsize=(10,6)); plt.barh(labels[::-1], scores[::-1]); plt.xlabel('freq x subtoken count'); plt.title('Top mined RimWorld tokens'); plt.tight_layout()
```
