# 03 — Replay a recorded episode
Load an episode JSON and step through its game states, reasoning, actions, and rewards.

```python
import sys
sys.path.insert(0, '..')
from pathlib import Path
from rimworld_agent.game.episode_recorder import load_episode
eps = sorted(Path('../data/episodes').glob('*/episode.json'))
print('episodes:', [p.parent.name for p in eps])
ep = load_episode(eps[0]) if eps else None
```

```python
from rimworld_agent.training.prepare_data import render_game_state
for step in (ep.steps[:3] if ep else []):
    print(render_game_state(step.game_state, step.visible_sids))
    print('REASONING:', step.reasoning)
    for a in step.actions:
        print('  ', a.action, a.params, '--', a.reason)
    print('REWARD:', round(step.reward.get('total', 0), 3)); print('='*60)
```

```python
import matplotlib.pyplot as plt
if ep:
    rewards = [s.reward.get('total', 0) for s in ep.steps]
    plt.plot(range(len(rewards)), rewards, marker='.'); plt.xlabel('step'); plt.ylabel('reward'); plt.title(ep.episode_id)
```
