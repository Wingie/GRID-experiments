```python
import sys, json
sys.path.insert(0, '..')
from pathlib import Path
data = json.loads(Path('../results/sid_assignments.json').read_text())
print('levels', data['levels'], 'codebook', data['codebook_size'], 'entities', data['num_entities'])
print('codebook usage per level:', data['codebook_usage'])
for a in list(data['assignments'].values())[:15]:
    print(f"{a['def_name'][:24]:24s} {a.get('category','?'):>9}/{a.get('subcategory','?'):<12} {a['sid_inline']}")
```

```python
# Project + colour by level-1 code or by entity-graph category.
from omegaconf import OmegaConf
from rimworld_agent.semantic_ids.assign_ids import run_pipeline
from rimworld_agent.semantic_ids.visualize import plot_clusters
cfg = OmegaConf.load('../configs/base.yaml')
res = run_pipeline(cfg)
labels = [res.graph.category_of(e.def_name) for e in res.entities]
plot_clusters(res.embeddings, labels, '../results/sid_clusters.png', method='pca')
```
